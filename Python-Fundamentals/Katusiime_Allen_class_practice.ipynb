```python
#functions

def bake_cake():
    print("milk")
    print("baking_soda")
    print("baking_powder")
bake_cake()
```

```python
def calculator(a,b):
  print(a+b)
  print(a-b)
  print(a*b)
  print(a/b)
calculator(10,20)
```

```python
def greet(name):
  print(f"hello {name}")
greet("mommy")
```

```python
#lambda function
add = lambda a,b:a+b
add(2,3)
```

```python
check_score=lambda score:"Pass"if score>=50 else "Fail"
print(check_score(70))
```

```python
num = lambda number:"positive"if number >0 else "negative"
print(num(4))
```

```python
temperature = lambda degree: "Has fever" if degree>37.5 else "cold" if degree<35 else "Normal temperature"
print(temperature(34))
```

```python
#recursion
#Recursion = function calling itself + a condition that eventually stops it.
def count(number):
  if number ==0:
    return
  print(number)

  count(number-1)
count(5)
```

```python
#print each score using recursion.

scores = [80, 65, 92, 74, 88]
def show_scores(scores,index):
  if index == len(scores):
    return
  print(scores[index])
  show_scores(scores,index+1)

show_scores(scores,0)
```
